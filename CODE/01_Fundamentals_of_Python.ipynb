## 🧮 **STEP 02: Variables in Python – Storing and Managing Data**

Variables are fundamental for data handling in drug discovery, acting as named containers for numbers, text, and complex structures. Understanding their use is key to writing effective and readable code.

### **Python Variable Naming Rules:**
To keep your code clean and functional, follow these rules when naming variables:
1.  **Start with a letter or an underscore:** `my_variable`, `_temp`
2.  **Can contain letters, numbers, and underscores:** `variable_1`, `data_point_2`
3.  **Case-sensitive:** `age` is different from `Age`.
4.  **Cannot be Python keywords:** Avoid words like `if`, `for`, `while`, `class`, `def`, `import`, etc.
5.  **Descriptive names are best:** Choose names that clearly indicate the variable's purpose (e.g., `patient_id` instead of `x`).
6.  **Cannot start with numbers:** `1_variable` is invalid.
7.  **Cannot contain special characters (except underscore):** `my-variable` is invalid, but `my_variable` is valid.

```python
# 01 Naming a Variable
num_1 = 30
num_2 = 20
sum = num_1 + num_2
```

```python
# 02 Seeing the Variable
print("Value of num_1 is", num_1)
print("Value of num_2 is", num_2)
print("Sum of num_1 and num_2 is", sum)
```

## 📝 **STEP 10: Working with Loops (For Loop) – Iterating Through Data**

In drug discovery, loops, particularly `for` loops, are essential for automating repetitive tasks, processing collections of data, and iterating through experimental results or molecular structures. They allow you to execute a block of code multiple times, once for each item in a sequence.

Understanding `for` loops is crucial for tasks like:
*   **Data Processing:** Applying a function or analysis to each element in a list of compounds or patient records.
*   **Simulations:** Running multiple iterations of a simulation.
*   **Bioinformatics:** Iterating over sequences of DNA, RNA, or proteins.

This section covers:
*   **Basic `for` loops:** Iterating over lists, tuples, and strings.
*   **`range()` function:** Generating sequences of numbers for iteration.
*   **Loop control statements:** Using `break` and `continue` to alter loop behavior.
*   **Practical Examples:** Demonstrating how `for` loops are used in drug discovery contexts.

```python
# 01 Understanding for loop using a list
pIC50 = [8.5, 7.8, 7.4, 4.44, 3.15,4.89, 6.5, 9.1]
```

```python
# Defining a for loop for compound labeling
bioactivity_class = []
for i in pIC50:
  if i >= 6.0:
    bioactivity_class.append("Active")
  elif i >= 4.50 and i < 6.0:
    bioactivity_class.append("Intermediate")
  else:
    bioactivity_class.append("Inactive")
```

```python
bioactivity_class
```

In drug discovery, functions are indispensable for organizing code, promoting reusability, and making complex programs more manageable. They allow you to encapsulate a specific task or a set of operations into a single block of code that can be called multiple times throughout your program.

Understanding functions is crucial for tasks like:

*   **Reusability:** Avoiding repetitive code by defining a task once and calling it whenever needed (e.g., calculating a specific molecular property).
*   **Maintainability:** Making your code easier to debug, update, and understand.

This section covers:
*   **Defining Functions:** How to create your own functions using the `def` keyword.
*   **Function Parameters:** Passing data into functions.
*   **Return Values:** Getting results back from functions.
*   **Scope:** Understanding local and global variables within functions.
*   **Practical Examples:** Demonstrating how functions can simplify tasks in drug discovery, such as calculating molecular descriptors or processing experimental data.

```python
# 01 Create a Basic Function
def my_func(name= "John"):
  print("Hello" + "" + name)
```

```python
my_func()
```

```python
# Function to convert nanomolar (nM) to molar (M)
def convert_nM_to_M(value_nM):
  """Converts a value from nanomolar (nM) to molar (M)."""
  return value_nM * 1e-9
```

```python
# Example list of IC50 values in nM
ic50_nM_values = [100, 500, 1000, 250, 750]

# Apply the function to convert each nM value to M
ic50_M_values = [convert_nM_to_M(val) for val in ic50_nM_values]

print("Original IC50 values (nM):", ic50_nM_values)
print("Converted IC50 values (M):", ic50_M_values)
```
